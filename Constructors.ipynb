```python
class Employee():

  def __init__(self,empname,empdept):
    self.empname=empname
    self.empdept=empdept

  def info(self):
    print(f"Employee {self.empname} works for {self.empdept}")

emp1=Employee("Imran","IT")
emp1.info()
```
