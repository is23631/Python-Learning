```python
class Employee():

  def __init__(self,empname,empdept):
    self.empname=empname
    self.empdept=empdept

  def info(self):
    print(f"Employee {self.empname} works for {self.empdept}")

emp1=Employee("Imran","IT")
emp1.info()
```

```python
class employee():

  company_name="Synechron"

  def __init__(self,empname,empdept):
    self.empname=empname
    self.empdept=empdept

  def changes(self,new_company):
    employee.company_name=new_company

  def info(self):
    print(f"Employee {self.empname} used to work in {self.empdept} for the company {employee.company_name}")

e1=Employee("Imran","IT")
e2=Employee("Priyansh","Finance")
e1.changes("Synechron India")
e2.info()

```
